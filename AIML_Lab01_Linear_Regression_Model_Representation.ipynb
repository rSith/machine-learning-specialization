```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
```

## Problem Statment
> Using a simple dataset with only two data points - a house with 1000 square feet(sqft) sold for \$300,000 and a house with 2000 sqft sold for $500,000.
These two points will constityte our data or training set.

```python
# x_train is the input variable (size in 1000 square feet)
# y_train is the target (price in 1000 of dollors)

x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
```

## Model Function
the model function for linear regression (which is a function that maps from `x` to `y`) is represented as

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b $$

The formula above is how you can represent straight lines - different values of $w$ and $b$ give you different straight lines on the plot.

```python
# Let's strat with w = 100 and b = 100
w = 100
b = 100

print(f"w: {w}")
print(f"b: {b}")
```

### Compute the value of $f_{w,b}(x^{(i)})$ for data points.
> for large number of data points calculate the function output in a `for` loop as shown in `compute_model_output` function
- `(ndarray(m,))` describes a Numpy n-dimensional array of shape m.
- `(scalar)` describes an argument without dimensions
- `np.zero(n)` will return a one-dimensional numpy array with n entries.

```python
"""Computes the prediction of a linear model
   Arguments :
   x (ndarray (m,)): Data, m examples
   w, b (scalar)  : model parameters
   Returns
   f_wb (ndarray (m,)): model prediction
"""

def compute_model_output(x, w, b):
  m = x.shape[0]
  f_wb = np.zeros(m)
  for i in range(m):
      f_wb[i] = w * x[i] + b

  return f_wb
```

```python
tmp_f_wb = compute_model_output(x_train, w, b)

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b', label='Our predctions')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual Values')

# Set the title
plt.title("Housing Prices")

# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')

# Set the x-axis label
plt.xlabel('Size (1000 sqft)')

plt.legend()
plt.show()
```

## Prediction
Now we have a model, we can use it to make our original prection.
> Let's predict the price of a house with 1200 sqft. (Since the units of $x$ are in 1000's of sqft, $x$ is 1.2)

```python
w = 200
b = 100
x_i = 1.2

cost_1200sqft = w * x_i + b

print(f"Price of a house with 1200 sqft = ${cost_1200sqft}")
```
